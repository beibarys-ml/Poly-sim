```python
import json
import os

with open(f"{FOP_ROOT}/speaker_map.json") as f:
    speaker_map = json.load(f)

items = list(speaker_map.items())
batch_size = 5

for i in range(0, len(items), batch_size):
    batch_id = i // batch_size
    batch = dict(items[i:i + batch_size])

    out_path = f"{FOP_ROOT}/speaker_map_batch_{batch_id:02d}.json"
    with open(out_path, "w") as f:
        json.dump(batch, f, indent=2)

    print(f"Batch {batch_id:02d}: {len(batch)} speakers")
```

```python
import os
import subprocess
from pathlib import Path

FOP_ROOT = "/content/drive/MyDrive/FOP"
VIDEO_ROOT = "/content/drive/MyDrive/data/videos"

batch_files = sorted(Path(FOP_ROOT).glob("speaker_map_batch_*.json"))

for batch_file in batch_files:
    batch_id = batch_file.stem.split("_")[-1]
    out_csv = f"{FOP_ROOT}/feature_tracker/v3_train_German_synthetic_batch_{batch_id}.csv"
    log_file = f"{FOP_ROOT}/pipeline_output/batch_{batch_id}.log"

    # Skip already completed batches
    if os.path.exists(out_csv) and os.path.getsize(out_csv) > 0:
        print(f"Skipping batch {batch_id}: CSV already exists")
        continue

    print(f"\n==============================")
    print(f"Running batch {batch_id}")
    print(f"Speaker map: {batch_file}")
    print(f"Output CSV: {out_csv}")
    print(f"Log file: {log_file}")
    print(f"==============================\n")

    cmd = [
        "python", "pipeline/generate_dataset.py",
        "--video_dir", VIDEO_ROOT,
        "--speaker_map", str(batch_file),
        "--src_language_label", "English",
        "--output_root", FOP_ROOT,
        "--tgt_lang", "de",
        "--tgt_language_label", "German",
        "--out_csv", out_csv,
        "--pipeline_output_dir", f"{FOP_ROOT}/pipeline_output",
        "--wav2lip_dir", f"{FOP_ROOT}/Wav2Lip",
        "--wav2lip_checkpoint", f"{FOP_ROOT}/Wav2Lip/checkpoints/wav2lip_gan.pth",
    ]

    with open(log_file, "w") as log:
        process = subprocess.run(
            cmd,
            stdout=log,
            stderr=subprocess.STDOUT,
            text=True,
        )

    if process.returncode != 0:
        print(f"Batch {batch_id} failed. Check log: {log_file}")
    else:
        print(f"Batch {batch_id} finished successfully.")
```
